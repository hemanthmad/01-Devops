```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_excel("sample-27.xlsx")
df['Combined'] = df['Time'].astype(str) + ' ' + df['AMPM']
df.drop(['Time', 'AMPM'], axis=1, inplace=True)
df




```

```python
top20 = df.nlargest(20, ['await'])[['DEV', 'await', "%util" , "Combined" , "tps" ,"rkB/s"]]
print(top20)
```

```python
df_filtered = df[df['await'] >= 2 ]
df_filtered

```

```python
df_filtered.plot(x='Combined', y=["%util" , "await"], kind='bar', title='Utilization', figsize=(20, 5))
plt.xticks(rotation=45)
plt.show()
```
